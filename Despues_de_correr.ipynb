```python
def establecer_tipo_persona_datos_faltantes(row):
    if row["TIPO_DOCUMENTO_IC"] in ["PASAP", "CC", "NUIP", "CE", "NREXT", "PPT", "RC", "CD", "TI", "C","P","T","E"]:
        return "PERSONA NATURAL"
    elif row["TIPO_DOCUMENTO_IC"] in ["NIT","N"]:
        return "PERSONA JURIDICA"
    else:
        return "DESCONOCIDO"
```
